```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from transformers import pipeline, AutoTokenizer, AutoModelForSequenceClassification
import torch
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import warnings
warnings.filterwarnings('ignore')
```

```python
# Download NLTK resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')
```

```python
print("Setting up sentiment analysis pipeline...")
# Load pre-trained sentiment analysis model from Hugging Face
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Create sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)

# Define a more advanced emotion classifier pipeline for nuanced emotions
emotion_model_name = "j-hartmann/emotion-english-distilroberta-base"
emotion_tokenizer = AutoTokenizer.from_pretrained(emotion_model_name)
emotion_model = AutoModelForSequenceClassification.from_pretrained(emotion_model_name)
emotion_classifier = pipeline("text-classification", model=emotion_model, tokenizer=emotion_tokenizer)
```

```python
# Function to clean and preprocess text
def preprocess_text(text):
    if not isInstance(text, str):
        return ''
    # Convert to lowercase
    text = text.lower()
    
    # Remove special characters, numbers, punctuations
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    # Remove extra whitespaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Tokenize text
    tokens = nltk.word_tokenize(text)
    
    # Remove stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    
    # Lemmatize tokens
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return ' '.join(tokens)

```

```python
def analyze_sentiment(text):
    result = sentiment_analyzer(text)[0]
    return result

def detect_emotion(text):
    # Using the return_all_scores parameter to get all emotion scores
    results = emotion_classifier(text, return_all_scores=True)[0]
    emotions = {item['label']: item['score'] for item in results}
    return emotions

def detect_specific_emotions(text):
    """Use keyword matching to detect specific emotions beyond what the model can identify"""
    text = text.lower()
    
    # Create emotion keyword dictionary
    emotion_keywords = {
        'joy': ['happy', 'joyful', 'excited', 'delighted', 'glad', 'pleased', 'content', 'elated', 'thrilled'],
        'sad': ['sad', 'unhappy', 'miserable', 'depressed', 'down', 'blue', 'gloomy', 'heartbroken', 'grief'],
        'angry': ['angry', 'furious', 'mad', 'irritated', 'annoyed', 'frustrated', 'outraged', 'resentful'],
        'anxious': ['anxious', 'worried', 'nervous', 'uneasy', 'concerned', 'stressed', 'tense', 'apprehensive'],
        'stressed': ['stressed', 'overwhelmed', 'pressure', 'burden', 'strain', 'tension', 'busy', 'overloaded'],
        'tired': ['tired', 'exhausted', 'sleepy', 'fatigued', 'drained', 'weary', 'lethargic', 'burned out'],
        'fearful': ['afraid', 'scared', 'frightened', 'terrified', 'fearful', 'panic', 'horror', 'dread'],
        'depressed': ['depressed', 'hopeless', 'worthless', 'empty', 'numb', 'despair', 'despondent']
    }
    
    # Check for matches
    detected_emotions = {}
    for emotion, keywords in emotion_keywords.items():
        for keyword in keywords:
            if keyword in text:
                detected_emotions[emotion] = detected_emotions.get(emotion, 0) + 1
    
    # Convert counts to scores (normalized)
    max_count = max(detected_emotions.values()) if detected_emotions else 1
    for emotion in detected_emotions:
        detected_emotions[emotion] = detected_emotions[emotion] / max_count
    
    return detected_emotions

def map_to_yoga_category(emotion_name, text="", sentiment_score=0):
    """Enhanced mapping that combines model output with text analysis"""
    
    # Check for specific emotion keywords in the text that might override the model
    keyword_emotions = detect_specific_emotions(text)
    
    # If we have strong keyword matches, they take precedence
    if keyword_emotions:
        strongest_keyword = max(keyword_emotions.items(), key=lambda x: x[1])
        if strongest_keyword[1] > 0.7:  # High confidence threshold
            return strongest_keyword[0]

            keyword_mapping = {
                'stressed': 'stress',
                'angry': 'anger', 
                'sad': 'sadness',
                'anxious': 'anxious',
                'tired': 'tired',
                'fearful': 'fear',
                'depressed': 'sadness',  # Map depression to sadness routines
                'joy': 'joy'
            }
            return keyword_mapping.get(strongest_keyword[0], 'neutral')
    # Base mapping from emotion model outputs to yoga routine categories
    emotion_map = {
        'joy': 'joy',
        'surprise': 'joy',
        'sadness': 'sadness',  # Not 'sad'
        'anger': 'anger',      # Not 'angry'
        'fear': 'fear',        # Not 'fearful'
        'disgust': 'anger',
        'neutral': 'neutral'
    }
    
    # Combine with sentiment analysis for better discrimination
    mapped_emotion = emotion_map.get(emotion_name, 'neutral')
    
    # Add additional logic based on sentiment model score
    if mapped_emotion == 'neutral':
        # For neutral classifications, check sentiment score to refine
        if sentiment_score > 0.9:  # Very positive sentiment
            return 'joy'
        elif sentiment_score < 0.1:  # Very negative sentiment
            # Check for more specific negative emotions in the text
            if any(word in text.lower() for word in ['stress', 'overwhelm', 'pressure']):
                return 'stressed'
            elif any(word in text.lower() for word in ['depress', 'hopeless', 'empty', 'meaningless']):
                return 'depressed'
            elif any(word in text.lower() for word in ['worry', 'anxious', 'nervous']):
                return 'anxious'
            else:
                return 'sad'  # Default negative emotion
    
    return mapped_emotion

```

```python



# Function to handle voice or text input (placeholder - will integrate with frontend)
def process_user_input(input_text):
    """Process user input to recommend appropriate yoga routines"""
    print(f"Analyzing input: '{input_text}'")
    
    # Step 1: Perform sentiment analysis
    sentiment = analyze_sentiment(input_text)
    sentiment_score = sentiment["score"]
    sentiment_label = sentiment["label"]
    
    # Step 2: Perform emotion detection
    emotions = detect_emotion(input_text)
    # Get the most likely emotion
    top_emotion = max(emotions.items(), key=lambda x: x[1])
    emotion_name = top_emotion[0]
    emotion_score = top_emotion[1]
    
    # Step 3: Detect specific emotions using keyword analysis
    keyword_emotions = detect_specific_emotions(input_text)
    
    # Print analysis results
    print(f"\nAnalysis Results:")
    print(f"Sentiment: {sentiment_label} ({sentiment_score:.4f})")
    print(f"Model detected emotion: {emotion_name} ({emotion_score:.4f})")
    
    if keyword_emotions:
        print("Keyword detected emotions:")
        for emotion, score in keyword_emotions.items():
            print(f"  - {emotion}: {score:.4f}")
    
    # Step 4: Map to appropriate yoga category using enhanced logic
    recommended_emotion = map_to_yoga_category(emotion_name, input_text, sentiment_score)
    
    print(f"\nFinal emotion category mapped to: {recommended_emotion}")
    
    # Step 5: Display recommendations
    if recommended_emotion in yoga_routines:
        routines = yoga_routines[recommended_emotion]
        print(f"\nBased on your mood (detected as: {recommended_emotion}), here are some recommended yoga routines:\n")
        
        for i, routine in enumerate(routines, 1):
            print(f"\n{i}. {routine['name']}")
            print(f"   Description: {routine['description']}")
            
            # Extract just the pose names from the pose dictionaries
            pose_names = [pose['name'] for pose in routine['poses']]
            print(f"   Key poses: {', '.join(pose_names)}")
            
            print(f"   Duration: {routine['duration']}")
            print(f"   Intensity: {routine['intensity']}")
            print(f"   Benefits: {routine['benefits']}")
    else:
        print(f"\nSorry, I don't have specific yoga recommendations for the emotion: {recommended_emotion}")
        print("Try expressing how you're feeling in different words, or choose from: joy, sadness, anger, anxiety, stress, fear, depression, or tiredness.")
```
